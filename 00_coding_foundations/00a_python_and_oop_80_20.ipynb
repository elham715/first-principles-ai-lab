## 1. Variables & Basic Data Types

Think of a **variable** as a labeled box holding a value in your computer's memory. You create one with `name = value` (no semicolon or type keyword needed).

### What the code below introduces:
* `int` — whole numbers (`epochs = 50`)
* `float` — decimal numbers (`lr = 0.001`)
* `str` — text wrapped in quotes (`name = "ElmanRNN"`)
* `bool` — truth values (`True` or `False`, capital first letter!)
* `None` — Python's explicit symbol for **"empty / nothing stored yet"**
* `f"..."` (**f-string**) — lets you inject variables directly inside `{}` inside text!
* `type(x)` — prints what kind of data is inside `x`.

## 2. Lists: Ordered Collections (`[...]`)

A **`list`** stores multiple values in order inside square brackets `[...]`. You can add items, check its length, or grab any slice.

### Critical Rules for Lists:
* **Zero-indexed:** The 1st item is at index `[0]`, the 2nd at `[1]`.
* **Negative indexing:** `[-1]` grabs the **last** item, `[-2]` the second-to-last (used constantly in RNNs to grab the final timestep!).
* **Slicing `[start:stop]`:** Includes `start`, **excludes** `stop`! So `nums[1:4]` grabs indices `1, 2, 3`.
* `.append(val)` adds an item to the end of the list.

> **Beginner Trap (Important for ML!):** A Python `list` is **NOT** a math vector! Doing `[1, 2] + [3, 4]` glues them into `[1, 2, 3, 4]`, and `[1, 2].mean()` crashes! For math, we convert lists to **NumPy arrays** (Notebook `00b`).

```python
losses = [2.4, 1.9, 1.3, 0.85, 0.42]

print("Total epochs recorded:", len(losses))
print("First loss (index 0):", losses[0])
print("Last loss (index -1):", losses[-1])
print("Middle slice [1:4]:", losses[1:4])     # indices 1, 2, 3

losses.append(0.31)
print("After append:", losses)
print("List + List just glues them:", [1, 2] + [3, 4])
```

## 3. Dictionaries (`{key: value}`) & Sets (`{a, b}`)

Two data structures appear in **every single ML/Bioinformatics project**:

1. **Dictionary (`dict`):** Maps a **unique key** to a **value**, written `{key: value}`.
   * **Why in ML?** Tokenizers! Mapping `"A" -> 0`, `"C" -> 1`, `"G" -> 2`, `"T" -> 3`, or storing hyperparameters.
   * `d.get(key, default)` safely returns `default` if the key is missing (prevents `KeyError` crashes).
2. **Set (`set`):** An **unordered collection of unique items** (duplicates are automatically deleted!).
   * **Why in Clinical ML?** Checking data leakage! If `train_patients` and `test_patients` are `set`s, `train_patients & test_patients` instantly tells you if any patient leaked into both splits!

```python
# Dictionary: amino-acid / DNA character -> integer token ID
vocab = {"A": 0, "C": 1, "G": 2, "T": 3}
vocab["<PAD>"] = 4  # add a new key-value pair

print("Token ID for G:", vocab["G"])
print("Safe lookup for X:", vocab.get("X", -1))  # returns -1 instead of crashing

# Set: unique patient IDs (duplicates removed automatically)
train_patients = set(["P001", "P002", "P003", "P001"])
test_patients = set(["P004", "P005"])

print("Unique train patients:", train_patients)
print("Any leakage between train & test?", train_patients & test_patients)  # empty set() = ZERO leakage!
```

## 4. Conditions (`if / elif / else`) & Loops (`for`, `enumerate`, `zip`)

Python uses **indentation (4 spaces)**—not curly braces `{}`—to know what code belongs inside an `if` or `for` block.

### 3 Loop Superpowers in ML:
1. **`for item in collection:`** loops over every item one by one.
2. **`enumerate(collection)`** gives you **both** the index `(0, 1, 2, ...)` AND the item—used in every epoch loop (`for epoch, loss in enumerate(losses):`).
3. **`zip(list_a, list_b)`** pairs items from two lists side-by-side (`for seq, label in zip(sequences, labels):`).

```python
sequences = ["ACGT", "TGCA", "GGCC"]
labels = [1, 0, 1]

# Condition check
accuracy = 0.86
if accuracy >= 0.90:
    status = "Strong"
elif accuracy >= 0.75:
    status = "Promising baseline"
else:
    status = "Needs tuning"
print("Model status:", status)

# Loop with enumerate + zip together
print("\nInspecting dataset samples:")
for idx, (seq, label) in enumerate(zip(sequences, labels)):
    tag = "Positive" if label == 1 else "Negative"
    print(f"  Sample {idx}: seq={seq} -> label={label} ({tag})")
```

## 5. List & Dict Comprehensions (Clean 1-Line Loops)

Instead of writing 4 lines to build a new list from an old one, Python lets you write a **comprehension** inside `[...]` or `{...}`.

### Read it from left to right:
`[expression  for item in collection  if condition]`

You will see this everywhere in **SMILES/peptide tokenizers** and **data preprocessing**.

```python
dna = "ACGTAC"
vocab = {"A": 0, "C": 1, "G": 2, "T": 3}

# 1. List comprehension: tokenize DNA string into integer IDs in 1 line
token_ids = [vocab[base] for base in dna]
print("Tokenized DNA:", token_ids)

# 2. Filtered list comprehension: keep only numbers > 0 (like ReLU!)
raw_vals = [-2.5, 0.0, 1.2, -0.4, 3.7]
positive_only = [v for v in raw_vals if v > 0]
print("Values > 0:", positive_only)

# 3. Dict comprehension: reverse the vocabulary (ID -> character) so we can decode predictions!
id_to_base = {idx: base for base, idx in vocab.items()}
print("Reversed vocab:", id_to_base)
```

## 6. Functions (`def`): Reusable Logic Blocks

A **function** takes inputs (parameters), runs code, and sends back an output with `return`.

### Syntax Rules You Must Know:
* Parameter syntax when **defining**: `name: type_hint = default_value`
  * `: float` is **only a note for humans** — Python does NOT convert types automatically!
  * `= 0.01` sets a **default** used if the caller omits that argument.
* When **calling**: write `compute_loss(2.0, 1.5)` or `compute_loss(y_true=2.0, y_pred=1.5)` — **never** write `: float` when calling!
* `*args` collects extra positional inputs as a tuple; `**kwargs` collects extra `key=value` inputs as a dictionary.

```python
def mse_loss(y_true: float, y_pred: float, scale: float = 1.0):
    error = (y_true - y_pred) ** 2
    return scale * error, abs(y_true - y_pred)  # returns 2 values as a tuple

# Call with positional values:
loss, mae = mse_loss(3.0, 2.5)
print(f"Loss={loss}, MAE={mae}")

# Call overriding the default `scale`:
scaled_loss, _ = mse_loss(y_true=3.0, y_pred=2.5, scale=0.5)
print(f"Scaled loss={scaled_loss}")

# *args and **kwargs demo (common in PyTorch wrappers)
def log_experiment(*metrics, **hyperparams):
    print("Metrics tuple:", metrics)
    print("Hyperparams dict:", hyperparams)

log_experiment(0.81, 0.89, lr=1e-3, hidden_dim=64)
```

## 7. OOP Core: `class`, `__init__`, and `self`

Functions forget everything once they finish. A **`class`** lets you bundle **memory (attributes)** + **actions (methods)** into a reusable blueprint.

### Understanding `self` in 10 Seconds:
* Suppose you build **two** counters from the same blueprint: `c1 = StepCounter()` and `c2 = StepCounter()`.
* How does `increment()` know whether to add `1` to `c1` or `c2`?
* Python silently converts `c1.increment()` into `StepCounter.increment(c1)`. Inside the method, **`self` IS `c1`**!
* Variables saved on `self.` (`self.count`) stay alive inside that specific object forever. Variables *without* `self.` vanish when the method finishes.

```python
class StepCounter:
    def __init__(self, name: str, start: int = 0):
        # `__init__` runs AUTOMATICALLY when you call StepCounter(...)
        self.name = name        # saved inside THIS object's memory
        self.count = start

    def increment(self, step: int = 1):
        # Even if you pass 0 args (`c1.increment()`), Python passes `self` automatically!
        self.count += step

    def report(self):
        return f"[{self.name}] current step = {self.count}"

train_counter = StepCounter("Train")
val_counter = StepCounter("Validation", start=10)

train_counter.increment()
train_counter.increment(4)
val_counter.increment()

# Notice both objects keep completely separate `self.count` values!
print(train_counter.report())
print(val_counter.report())
```

## 8. Dunder (Magic) Methods: `__call__`, `__len__`, `__getitem__`, `__repr__`

Methods surrounded by double underscores (`__name__`) connect your custom classes to **built-in Python operators**:
* `__call__(self, x)` → lets you run `obj(x)` just like calling a function! (**PyTorch `nn.Module` uses this!**)
* `__len__(self)` → lets you run `len(obj)`! (**PyTorch `Dataset` requires this!**)
* `__getitem__(self, idx)` → lets you run `obj[idx]`! (**PyTorch `Dataset` requires this!**)
* `__repr__(self)` → controls what `print(obj)` displays!

```python
class MiniSequenceDataset:
    def __init__(self, sequences: list, labels: list):
        self.sequences = sequences
        self.labels = labels

    def __len__(self) -> int:
        return len(self.sequences)

    def __getitem__(self, idx: int):
        return self.sequences[idx], self.labels[idx]

    def __repr__(self) -> str:
        return f"MiniSequenceDataset(n_samples={len(self)})"

class Multiplier:
    def __init__(self, factor: float):
        self.factor = factor

    def __call__(self, x: float) -> float:
        return x * self.factor

ds = MiniSequenceDataset(["ACGT", "TTAA", "CCGG"], [1, 0, 1])
print(ds)                   # calls __repr__
print("Length:", len(ds))   # calls __len__
print("Item 1:", ds[1])     # calls __getitem__

scale = Multiplier(3.0)
print("scale(5) =", scale(5.0))  # calls __call__ -> 15.0
```

## 9. Inheritance, `super().__init__()`, and Method Overriding

**Inheritance** lets a child class reuse everything from a parent class and specialize only what changed.

### Critical Rules:
1. Write `class Child(Parent):` to inherit from `Parent`.
2. **ALWAYS call `super().__init__(...)`** inside the child's `__init__`! If you forget `super().__init__()`, the parent never initializes its variables (this is the **#1 PyTorch bug** when beginners subclass `nn.Module`!).
3. Defining a method with the same name in the child **overrides** (replaces or extends) the parent's version.

```python
class BaseModule:
    def __init__(self, name: str):
        self.name = name
        self.training = True

    def __call__(self, x: float) -> float:
        # Just like PyTorch nn.Module: calling model(x) routes to forward(x)!
        return self.forward(x)

    def forward(self, x: float) -> float:
        raise NotImplementedError("Child class must define forward()!")

class ReLUModule(BaseModule):
    def __init__(self):
        super().__init__(name="ReLU")   # initializes self.name & self.training in BaseModule!

    def forward(self, x: float) -> float:
        return x if x > 0 else 0.0      # overrides BaseModule.forward()

relu = ReLUModule()
print(f"Module={relu.name} | training={relu.training}")
print("relu(-3.5) =", relu(-3.5))   # calls __call__ -> forward()
print("relu(4.2)  =", relu(4.2))
```

## 10. Composition ("Has-A" Architecture — Building Model Pipelines)

Instead of cramming 10 layers into one giant class, **Composition** means a container class **holds smaller objects inside itself** (e.g., `self.layers = [...]`).  
This is exactly how neural networks are built: a `Model` *has* layers (`Conv2d`, `LSTM`, `Linear`).

```python
class AddBias(BaseModule):
    def __init__(self, bias: float):
        super().__init__(name=f"AddBias({bias})")
        self.bias = bias

    def forward(self, x: float) -> float:
        return x + self.bias

class Scale(BaseModule):
    def __init__(self, weight: float):
        super().__init__(name=f"Scale({weight})")
        self.weight = weight

    def forward(self, x: float) -> float:
        return x * self.weight

class Pipeline(BaseModule):
    def __init__(self, layers: list):
        super().__init__(name="Pipeline")
        self.layers = layers  # Composition: Pipeline HAS a list of layer objects

    def forward(self, x: float) -> float:
        for layer in self.layers:
            x = layer(x)      # Polymorphism: every layer works via layer(x)!
        return x

net = Pipeline([
    Scale(2.0),
    AddBias(-5.0),
    ReLUModule(),
])

print("Input 1.0 ->", net(1.0))  # 1*2 - 5 = -3 -> ReLU -> 0.0
print("Input 4.0 ->", net(4.0))  # 4*2 - 5 =  3 -> ReLU -> 3.0
```

## 11. Modern Clean OOP: `@dataclass`, `@property`, `@classmethod`, `@staticmethod`, and `ABC`

These 5 decorators appear constantly in clean research codebases:
* **`@dataclass`**: Eliminates boilerplate `__init__` for configuration classes.
* **`@property`**: Makes a computed method look like a variable (`obj.area`, no `()`).
* **`@classmethod`**: Receives `cls` (the class itself) to create alternative constructors (`Model.from_config(cfg)`).
* **`@staticmethod`**: A pure helper function living inside a class (receives neither `self` nor `cls`).
* **`ABC` + `@abstractmethod`**: Refuses to let anyone forget required methods in child classes.

```python
from dataclasses import dataclass
from abc import ABC, abstractmethod

@dataclass
class HyperParams:
    lr: float = 1e-3
    hidden_dim: int = 64

class AbstractBioEncoder(ABC):
    def __init__(self, hidden_dim: int, lr: float):
        self.hidden_dim = hidden_dim
        self.lr = lr

    @property
    def doubled_dim(self) -> int:
        return self.hidden_dim * 2          # accessed WITHOUT ()

    @staticmethod
    def clean_seq(raw: str) -> str:
        return raw.strip().upper()          # pure helper utility

    @classmethod
    def from_hparams(cls, cfg: HyperParams):
        return cls(hidden_dim=cfg.hidden_dim, lr=cfg.lr)  # alternative constructor

    @abstractmethod
    def encode(self, seq: str) -> list:
        pass

class PeptideEncoder(AbstractBioEncoder):
    def encode(self, seq: str) -> list:
        cleaned = self.clean_seq(seq)
        return [ord(ch) for ch in cleaned]

cfg = HyperParams(lr=5e-4, hidden_dim=128)
enc = PeptideEncoder.from_hparams(cfg)
print("Config:", cfg)
print("Doubled dim (@property):", enc.doubled_dim)
print("Encoded ' acde ':", enc.encode(" acde "))
```
